```python
import ultralytics

print(ultralytics.__version__)
```

```python
from pathlib import Path
from ultralytics import YOLO
```

```python
DATA_YAML = Path("/kaggle/input/datasets/trungnghi/processed-ppe-monitoring-system/processed/data.yaml")
print(DATA_YAML.read_text())
```

```python
model = YOLO("yolo26s.pt")

results = model.train(
    data=DATA_YAML,
    epochs=100,
    imgsz=640,
    batch=-1,
    patience=25,
    project="/kaggle/working/runs",
    name="ppe_yolo26",
    device=0,
    cache=True
)
```
